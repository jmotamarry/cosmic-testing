```python
import os
import json
import pandas as pd
from seticore import viewer as seticore_viewer

# 🔧 Set your root directory here
root_dir = "/datax/scratch/jaym/meerkat_data"

# 📂 Recursively collect .hits file paths
def get_hits_filepaths(directory):
    hits = []
    for root, _, files in os.walk(directory):
        hits += [os.path.join(root, f) for f in files if f.endswith(".hits")]
    return hits

# 📥 Initialize data dictionary
d = {
    "file_path": [],
    "hit_file_enumeration": [],
    "signal_frequency": [],
    "signal_index": [],
    "signal_driftSteps": [],
    "signal_driftRate": [],
    "signal_snr": [],
    "signal_coarseChannel": [],
    "signal_numTimesteps": [],
    "signal_power": [],
    "signal_incoherentPower": [],
    "sourceName": [],
    "fch1": [],
    "foff": [],
    "tstart": [],
    "tsamp": [],
    "ra": [],
    "dec": [],
    "telescopeId": [],
    "numTimesteps": [],
    "numChannels": [],
    "coarseChannel": [],
    "startChannel": [],
    "beam": [],
}

hits_count = 0
hits_files = get_hits_filepaths(root_dir)
print(f"🧪 Found {len(hits_files)} .hits files")

# 🛠️ Process hits
for hits_filepath in hits_files:  # ← change [:3] to [:10] or remove for full run
    print(f"📂 Reading: {hits_filepath}")
    try:
        for hit_enum, hit in enumerate(seticore_viewer.read_hits(hits_filepath)):
            hits_count += 1
            d["file_path"].append(hits_filepath)
            d["hit_file_enumeration"].append(hit_enum)

            s = hit.signal
            f = hit.filterbank

            d["signal_frequency"].append(getattr(s, "frequency", None))
            d["signal_index"].append(getattr(s, "index", None))
            d["signal_driftSteps"].append(getattr(s, "driftSteps", None))
            d["signal_driftRate"].append(getattr(s, "driftRate", None))
            d["signal_snr"].append(getattr(s, "snr", None))
            d["signal_coarseChannel"].append(getattr(s, "coarseChannel", None))
            d["signal_numTimesteps"].append(getattr(s, "numTimesteps", None))
            d["signal_power"].append(getattr(s, "power", None))
            d["signal_incoherentPower"].append(getattr(s, "incoherentPower", None))

            d["sourceName"].append(getattr(f, "sourceName", None))
            d["fch1"].append(getattr(f, "fch1", None))
            d["foff"].append(getattr(f, "foff", None))
            d["tstart"].append(getattr(f, "tstart", None))
            d["tsamp"].append(getattr(f, "tsamp", None))
            d["ra"].append(getattr(f, "ra", None))
            d["dec"].append(getattr(f, "dec", None))
            d["telescopeId"].append(getattr(f, "telescopeId", None))
            d["numTimesteps"].append(getattr(f, "numTimesteps", None))
            d["numChannels"].append(getattr(f, "numChannels", None))
            d["coarseChannel"].append(getattr(f, "coarseChannel", None))
            d["startChannel"].append(getattr(f, "startChannel", None))
            d["beam"].append(getattr(f, "beam", None))

    except Exception as e:
        print(f"⚠️ Failed to parse {hits_filepath}: {e}")

# 💾 Save to JSON
out_json = os.path.join(root_dir, "hits_collation.json")
with open(out_json, "w") as f:
    json.dump(d, f)

# 🧾 Load into DataFrame
df = pd.DataFrame(d)
print(f"✅ {hits_count} hits collected from {len(hits_files)} files")
print(f"📄 JSON saved to: {out_json}")
print(df.head())

# Convert dict to DataFrame
df = pd.DataFrame(d)

# Save to pickle for fast loading later
pkl_path = root_dir + '/../meerkat_hits.pkl'
df.to_pickle(pkl_path)

print(f"💾 Pickle saved to: {pkl_path}")
```
