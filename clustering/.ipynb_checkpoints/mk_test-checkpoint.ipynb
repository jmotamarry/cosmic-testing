```python
# --- MULTI-ANTENNA AUTOENCODER PIPELINE ---

import numpy as np
import glob
import os
import matplotlib.pyplot as plt
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import (
    InputLayer, Conv2D, MaxPooling2D, Flatten, Dense,
    Reshape, Conv2DTranspose, BatchNormalization, Activation, Input
)
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
from collections import defaultdict
from sklearn.cluster import HDBSCAN
import umap

# --- STEP 0: Inject synthetic ET signals ---
def inject_synthetic_signal(stamp, freq_center=None, drift_rate=None, snr=None):
    t, f, a = stamp.shape
    if freq_center is None:
        freq_center = np.random.randint(50, f - 50)
    if drift_rate is None:
        drift_rate = np.random.uniform(2.0, 10.0)
    if snr is None:
        snr = np.random.uniform(15.0, 50.0)
    slope = drift_rate / t
    selected_ants = np.random.choice(a, size=np.random.randint(2, 6), replace=False)
    for ant in selected_ants:
        for ti in range(t):
            fi = int(freq_center + slope * ti)
            if 0 <= fi < f:
                stamp[ti, fi, ant] += snr * np.std(stamp)
    return stamp

# --- STEP 1: Load (T, F, P, A) stamps ---
STAMPS_GLOB = "/datax/scratch/jaym/resampled_mk_stamps/*.npy"
paths = glob.glob(STAMPS_GLOB)
stamps = []

for path in paths:
    try:
        arr = np.load(path)  # shape: (T, F, P, A)
        stamps.append(arr)
    except Exception as e:
        print(f"Error loading {path}: {e}")

print(f"Loaded {len(stamps)} stamp files.")

# --- STEP 2: Convert to (T, F, A) by summing over pol ---
def get_full_stamp(arr):
    if arr.shape[2] > 1:
        arr = arr.sum(axis=2)
    else:
        arr = arr[:, :, 0, :]
    return arr

# --- STEP 2.5: Pad or crop with chi-squared noise to ensure uniform shape ---
TARGET_TIME = 16
TARGET_FREQ = 640
TARGET_ANTS = 64
STAMP_SHAPE = (TARGET_TIME, TARGET_FREQ, TARGET_ANTS)

synthetic_indices = set()

def pad_or_crop_stamp(a, target_shape=STAMP_SHAPE, inject_synthetic=False):
    t, f, a_dim = a.shape
    t_pad = max(0, target_shape[0] - t)
    f_pad = max(0, target_shape[1] - f)
    a_pad = max(0, target_shape[2] - a_dim)

    pad_width = ((0, t_pad), (0, f_pad), (0, a_pad))
    padded = np.pad(a, pad_width, mode='constant', constant_values=0)
    cropped = padded[:target_shape[0], :target_shape[1], :target_shape[2]]

    if t_pad > 0 or f_pad > 0 or a_pad > 0:
        signal_mean = np.mean(a)
        noise = np.random.chisquare(df=2, size=STAMP_SHAPE) * signal_mean / 2
        mask = (padded == 0)
        cropped[mask] = noise[mask]

    if inject_synthetic:
        cropped = inject_synthetic_signal(cropped)

    return cropped

full_stamps = []
synthetic_mask = []
for i, s in enumerate(stamps):
    try:
        full_arr = get_full_stamp(s)
        inject = (i % 25 == 0)
        synthetic_mask.append(inject)
        if inject:
            synthetic_indices.add(len(full_stamps))
        full_arr = pad_or_crop_stamp(full_arr, inject_synthetic=inject)
        # Stack antennas vertically: reshape (T, F, A) → (T*A, F)
        full_arr = full_arr.transpose(2, 0, 1).reshape(-1, full_arr.shape[1])
        full_stamps.append(full_arr[:, :, np.newaxis])  # Add channel dimension
    except Exception as e:
        print(f"Failed processing stamp {i}: {e}")

# --- STEP 3: Normalize ---
X = np.stack(full_stamps, axis=0)
X_norm = np.empty_like(X)
for i in range(X.shape[0]):
    vmin, vmax = X[i].min(), X[i].max()
    if vmax > vmin:
        X_norm[i] = (X[i] - vmin) / (vmax - vmin)
    else:
        X_norm[i] = X[i]

X_input = X_norm
synthetic_mask = np.array(synthetic_mask)

# --- STEP 4: Split (exclude synthetic from training) ---
X_train, X_val = train_test_split(X_input[~synthetic_mask], test_size=0.2, random_state=42)

# --- STEP 5: Build Autoencoder ---
input_shape = X_input.shape[1:]
latent_dim = 128

model = Sequential([
    InputLayer(input_shape=input_shape),
    Conv2D(32, (3, 3), padding='same'), BatchNormalization(), Activation('relu'),
    MaxPooling2D((2, 2), padding='same'),
    Conv2D(64, (3, 3), padding='same'), BatchNormalization(), Activation('relu'),
    MaxPooling2D((2, 2), padding='same'),
    Flatten(),
    Dense(latent_dim, name="latent_vector"),
    Dense((input_shape[0] // 4) * (input_shape[1] // 4) * 64),
    Reshape((input_shape[0] // 4, input_shape[1] // 4, 64)),
    Conv2DTranspose(64, (3, 3), strides=(2, 2), padding='same'), BatchNormalization(), Activation('relu'),
    Conv2DTranspose(32, (3, 3), strides=(2, 2), padding='same'), BatchNormalization(), Activation('relu'),
    Conv2D(input_shape[2], (3, 3), activation='sigmoid', padding='same')
])

model.compile(optimizer='adam', loss='mse')
model.build(input_shape=(None,) + input_shape)
_ = model.predict(X_input[:1])
model.summary()

# --- STEP 6: Train ---
history = model.fit(
    X_train, X_train,
     epochs=5,
    batch_size=16,
    validation_data=(X_val, X_val),
    callbacks=[EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)]
)

# --- STEP 7: Plot Training Loss ---
plt.plot(history.history['loss'], label='Train')
plt.plot(history.history['val_loss'], label='Val')
plt.title("Loss Curve")
plt.xlabel("Epoch")
plt.ylabel("MSE Loss")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# --- STEP 8: Latent Vectors & MSE ---
reconstructions = model.predict(X_input, batch_size=16)
mse_per_sample = np.mean((X_input - reconstructions) ** 2, axis=(1, 2, 3))

encoder_input = Input(shape=input_shape)
encoded = encoder_input
for layer in model.layers:
    encoded = layer(encoded)
    if layer.name == "latent_vector":
        break
encoder = Model(inputs=encoder_input, outputs=encoded)
latent_vectors = encoder.predict(X_input, batch_size=16)

# --- STEP 9: HDBSCAN Clustering with UMAP Visualization ---
def clustering_with_visualization(latent_vectors, round_number):
    reducer = umap.UMAP(n_components=2, random_state=round_number)
    latent_2d = reducer.fit_transform(latent_vectors)
    clusterer = HDBSCAN(min_samples=5, min_cluster_size=5)
    labels = clusterer.fit_predict(latent_vectors)
    plt.figure(figsize=(10, 6))
    plt.scatter(latent_2d[:, 0], latent_2d[:, 1], c=labels, cmap='Spectral', s=5)
    plt.title(f"Latent Space after HDBSCAN Round {round_number}")
    plt.colorbar(label="Cluster Label")
    plt.tight_layout()
    plt.show()
    return labels

labels = clustering_with_visualization(latent_vectors, round_number=1)
outlier_indices = np.where(labels == -1)[0]
print(f"Outliers from HDBSCAN: {len(outlier_indices)}")

# --- STEP 10: Visualize outliers ---
for idx in outlier_indices:
    original = X_input[idx][:, :, 0].reshape(TARGET_ANTS, TARGET_TIME, TARGET_FREQ).transpose(1, 2, 0)
    fig, axes = plt.subplots(4, 16, figsize=(20, 5))
    for ant in range(TARGET_ANTS):
        ax = axes[ant // 16, ant % 16]
        ax.imshow(original[:, :, ant], aspect='auto', origin='lower', cmap='inferno')
        ax.set_title(f"Ant {ant}", fontsize=6)
        ax.axis('off')
    plt.suptitle(f"Final Anomalous Stamp #{idx}" + (" [SYNTHETIC]" if idx in synthetic_indices else ""))
    plt.tight_layout()
    plt.show()
```
